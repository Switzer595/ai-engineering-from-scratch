## Exercise 1: Compare list and NumPy performance

使用 `%timeit` 比较 Python list 和 NumPy 在创建 10 万个随机数数组时的性能。

```python
import random
import numpy as np
```

```python
%timeit [random.random() for _ in range(100_000)]
```

```python
%timeit np.random.random(100_000)
```

结论：NumPy 创建大量随机数通常比 Python 原生 list 更快，因为 NumPy 底层使用更高效的数组实现。
